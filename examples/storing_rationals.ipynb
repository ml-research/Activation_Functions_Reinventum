# Training new rational initialization weights
A rational can be initiaized such that it approximates any function. To save computation the weights can be pretrained and reused for different rationals by storing them in a json file.

```python
import torch
import torch.nn as nn

from activations.torch import ActivationModule, rationals
from activations.utils import JsonHandler

JsonHandler.create("./example.json", copy=False)
```

First we train a rational such that it approximates ``ReLU`` using ``rationals.find_weights`` method.

```python
relu = nn.ReLU()
w_numerator, w_denominator = rationals.find_weights(
    relu,  # the function to approximate
    torch.linspace(torch.tensor(-3.), torch.tensor(3.), 100),  # range to approximate on
    (5, 4),  # degree of rational
    "B"  # version of rational
)
```

Now store the weights in a json file.

```python
JsonHandler.store(
    version="A",
    degrees=(5, 4),
    name="relu",  # the name under which weights are stored
    numerator=w_numerator,
    denominator=w_denominator,
)
```
