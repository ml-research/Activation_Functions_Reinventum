# Using names and groups arguments
Each module must be regstered under a unique name. This name will be used to reference the module througout the library.
Optionally a module can be assigned to one or more groups. A group is a collection of modules under a unique name.

```python
import torch
import torch.nn as nn

from activations.torch import ActivationModule
```

First, lets register some activations

```python
relu = nn.ReLU()
ActivationModule.register(relu, "relu", group="group_1")

sigmoid = nn.Sigmoid()
ActivationModule.register(sigmoid, "sigmoid", group=["group_1", "group_2", "group_3"])

tanh = nn.Tanh()
ActivationModule.register(tanh, "tanh", group="group_3")
```

```python
for i in range(3):
    module_names = ActivationModule.get_groups(f"group_{i+1}")
    print(f"group_{i+1}: {module_names}")
```

Now we can reference multiple activations via ``group`` argument. More than one group can be passed as argument and modules belonging to multiple given groups will only be referenced once.

```python
ActivationModule.create_snapshot(group=None, function=True)  # references all modules

ActivationModule.show_function(group="group_1", function=True, title="group_1")
```

```python
ActivationModule.show_function(group=["group_1", "group_3"], function=True, title="group_1, group_3")
```
