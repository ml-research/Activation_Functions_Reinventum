# Using custom activations for capturing/plotting

```python
import torch
import torch.nn as nn

from activations.torch import ActivationModule
```

To register any activation (more specifically any ``nn.Module``) use the ``ActivationModule.register`` method.

```python
relu = nn.ReLU()

registered_name = ActivationModule.register(relu, "my_relu")
print(f"Activation is registered under {registered_name}")
```

If a name is not unique '_`i`' is appended. If such a suffix is already present it will be incremented.

```python
relu2 = nn.ReLU()
relu3 = nn.ReLU()

ActivationModule.register(relu2, "my_relu")  # suffix '_0' is appended
ActivationModule.register(relu3, "my_relu_0")  # suffix will be incremented to '_1'

print(ActivationModule.modules())
```

To register instances of a custom activation class, call `ActivationModule.register(self,...)`.

```python
class CustomActivation(nn.Module):

    def __init__(self):
        super().__init__()  # initialize ``nn.Module``

        ActivationModule.register(self, "custom_activation_0")
```

```python
my_act1 = CustomActivation()
print(ActivationModule.modules())
```
